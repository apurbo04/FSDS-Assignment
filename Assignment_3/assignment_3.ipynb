1. Inventory Price Calculator

```python
def cis(*price):
    total = 0
    count = 0
    for i in price:
        total += i
        count += 1
    return total, total / count
```

```python
result = cis(120, 250, 399, 150)
print(result)
```

2. Unique Visitor Tracker

```python
def cuv(visitor_list):
    u_list = set(visitor_list)           
    return len(u_list), u_list           
```

```python
visitor = [101, 205, 101, 310, 205, 550]
count, unique_ids = cuv(visitor)
print(f"{count} (The unique IDs are {unique_ids})")
```

3. Product Category Counter

```python
def count_categories(products):
    category_count = {}           

    for item in products:
        if item in category_count:
            category_count[item] += 1   
        else:
            category_count[item] = 1    

    return category_count
```

```python
products = ["Laptop", "Smartphone", "Monitor", "Laptop",
            "Smartphone", "Headphones", "Laptop"]

print(count_categories(products))
```

4. Temperature Analytics

```python
def analyze_temperatures(temp_list):
    high_list = []     
    normal_list = []    

    for temp in temp_list:
        if temp > 38:
            high_list.append(temp)
        else:
            normal_list.append(temp)

    return high_list, normal_list
```

```python
temps = [36.5, 38.0, 39.1, 37.2, 40.5, 38.0, 37.9]

high, normal = analyze_temperatures(temps)

print("High List (>38°C):", high)
print("Normal List (≤38°C):", normal)
print("Full Output Tuple:", (high, normal))
```

5. Menu Price Lookup

```python
def get_price(menu, item):
    if item in menu:
        return menu[item]            
    else:
        return "Item not found"       
```

```python
# Example 1: Item Found
menu = {"Coffee": 3.50, "Tea": 2.75, "Muffin": 4.00, "Sandwich": 8.50}
print(get_price(menu, "Muffin"))

# Example 2: Item Not Found
print(get_price(menu, "Donut"))
```

6. Student Grading System

```python
def calculate_grade(score):
    if score >= 80:
        return "A+"
    elif score >= 75:
        return "A"
    elif score >= 60:
        return "A-"
    else:
        return "F"

def grade_students(student_dict):
    grade_dict = {}
    for student, score in student_dict.items():
        grade_dict[student] = calculate_grade(score)
    return grade_dict
```

```python
scores = {"Alice": 85, "Bob": 78, "Charlie": 32, "David": 69}
print(grade_students(scores))
```

7. Discount Calculator

```python
def apply_discounts(products):
    updated_prices = []

    for name, price, discount in products:
        new_price = price - (price * discount / 100)
        updated_prices.append(round(new_price, 2))     

    return updated_prices
```

```python
items = [("Keyboard", 80.00, 20),
         ("Mouse", 25.00, 10),
         ("Monitor", 300.00, 5)]

print(apply_discounts(items))
```

8. Hashtag Generator

```python
def generate_hashtags(sentence):
    words = sentence.split()
    hashtags = ["#" + word.lower() for word in words]
    return hashtags
```

```python
print(generate_hashtags("AI Data Science"))
```

10. Shopping Cart Bill Calculator

```python
def calculate_total(cart, prices):
    total = 0.0
    for item in cart:
        if item in prices:
            total += prices[item]   
        else:
            total += 0              
    return round(total, 2)          
```

```python
cart = ["Milk", "Bread", "Milk", "Cheese", "Bread"]
prices = {"Milk": 3.00, "Bread": 2.50, "Eggs": 4.00, "Cheese": 5.00}

print(calculate_total(cart, prices))
```
