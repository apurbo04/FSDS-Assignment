1. Email Validator (String Filtering + Loops)

```python
email = input("Enter your email: ")

has_at = False
valid_end = False

for a in email:
    if a == '@':
        has_at = True
        break

if email.endswith(".com")or email.endswith(".org"):
    valid_end = True

if has_at and valid_end:
    print("Valid Email")
else:
    print("Invalid Email")
```

2. Count Word Frequency in Customer Reviews

```python
given_text = input("Enter your text: ")

given_text = given_text.lower()

words = given_text.split()
freq = {}

for w in words:
  if w in freq:
    freq[w] += 1
  else:
    freq[w] = 1

print(freq)
```

3. Password Strength Checker

```python
password = input("Enter password: ")

has_upper = False
has_lower = False
has_digit = False
has_special = False

special_chars = "!@#$%^&*()-_=+{}[]|;:'\",.<>?/`~"

for ch in password:
    if ch.isupper():
        has_upper = True
    elif ch.islower():
        has_lower = True
    elif ch.isdigit():
        has_digit = True
    elif ch in special_chars:
        has_special = True

# Final decision
if has_upper and has_lower and has_digit and has_special:
    print("Strong Password")
else:
    print("Weak Password")
```

4. Username Generator

```python
import random

full_name = input("Enter full name: ")

full_name = full_name.lower()

parts = full_name.split()
first = parts[0]
last = parts[-1]

first_part = ""
for i in range(3):
    if i < len(first):
        first_part += first[i]

last_part = ""
for i in range(len(last) - 3, len(last)):
    if i >= 0:
        last_part += last[i]

num = random.randint(10, 99)

username = first_part + last_part + str(num)
print(username)
```

5. SMS Word Counter

```python
message = input("Enter message: ")

char_count = 0
for ch in message:
    char_count += 1

words = message.split()
word_count = 0

for w in words:
    word_count += 1

if char_count > 160:
    status = "Exceeds Limit"
else:
    status = "Within Limit"

print("Words =", word_count)
print("Characters =", char_count)
print("Status =", status)
```

6. Remove Duplicates from Contact List

```python
names = input("Enter names: ")

items = names.split(",")
unique = []

for x in items:
    x = x.strip()
    if x not in unique:
        unique.append(x)

print(", ".join(unique))
```

7. Detect Suspicious Words in Text (Moderation)

```python
text = input("Enter text: ")
banned_words = ["spam", "scam", "fake"]

warning = False

for word in banned_words:
    if text.find(word) != -1:
        warning = True
        break

if warning:
    print("Warning: Suspicious content detected")
else:
    print("Text is clean")
```

8. Invoice Code Formatter

```python
num = int(input("Enter invoice number: "))

num_str = str(num)

while len(num_str) < 5:
    num_str = "0" + num_str

invoice_code = "INV" + num_str
print(invoice_code)
```

9. Count Hashtags and Mentions (Social Media Analytics)

```python
post = input("Enter post: ")

hashtag_count = 0
mention_count = 0

for ch in post:
    if ch == "#":
        hashtag_count += 1
    elif ch == "@":
        mention_count += 1

print("Hashtags =", hashtag_count)
print("Mentions =", mention_count)
```

10. Word Reversal for Data Encryption

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
reversed_words = []

for word in words:
    rev = ""
    for ch in word:
        rev = ch + rev
    reversed_words.append(rev)

result = " ".join(reversed_words)
print(result)
```
